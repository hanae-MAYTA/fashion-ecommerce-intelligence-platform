```python
from pathlib import Path

DATA_PATH = Path("../data/raw")

customers = pd.read_csv(DATA_PATH / "customers.csv")
discounts = pd.read_csv(DATA_PATH / "discounts.csv")
employees = pd.read_csv(DATA_PATH / "employees.csv")
products = pd.read_csv(DATA_PATH / "products.csv")
stores = pd.read_csv(DATA_PATH / "stores.csv")
transactions = pd.read_csv(DATA_PATH / "transactions.csv")
```

```python
def dataset_info(df, name):
    print(f"\n{'=' * 60}")
    print(f"{name}")
    print(f"{'=' * 60}")
    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")
    print("\nData types:")
    print(df.dtypes)
```

```python
def missing_values(df, name):
    result = pd.DataFrame({
        "Missing": df.isna().sum(),
        "Missing %": (df.isna().mean() * 100).round(2)
    })
    
    result = result[result["Missing"] > 0]
    
    print(f"\n{name}")
    display(result.sort_values("Missing", ascending=False))
```

```python
missing_values(customers, "CUSTOMERS")
```

```python
missing_values(discounts, "DISCOUNTS")
```

```python
missing_values(employees, "EMPLOYEES")
```

```python
missing_values(products, "PRODUCTS")
```

```python
missing_values(stores, "STORES")
```

```python
missing_values(transactions, "TRANSACTIONS")
```

```python
def duplicate_info(df, name):
    duplicates = df.duplicated().sum()
    print(f"{name}: {duplicates} duplicated rows")

duplicate_info(customers, "CUSTOMERS")
duplicate_info(discounts, "DISCOUNTS")
duplicate_info(employees, "EMPLOYEES")
duplicate_info(products, "PRODUCTS")
duplicate_info(stores, "STORES")
duplicate_info(transactions, "TRANSACTIONS")
```

```python
key_checks = {
    "Customer ID": customers["Customer ID"],
    "Product ID": products["Product ID"],
    "Store ID": stores["Store ID"],
    "Employee ID": employees["Employee ID"],
    "Invoice ID": transactions["Invoice ID"]
}

for key, series in key_checks.items():
    print(
        f"{key}: "
        f"{series.nunique()} unique values / "
        f"{len(series)} rows"
    )
```

```python
customer_ids = set(customers["Customer ID"])

missing_customers = transactions[
    ~transactions["Customer ID"].isin(customer_ids)
]

print("Transactions with unknown Customer ID:",
      len(missing_customers))
```

```python
product_ids = set(products["Product ID"])

missing_products = transactions[
    ~transactions["Product ID"].isin(product_ids)
]

print("Transactions with unknown Product ID:",
      len(missing_products))
```

```python
store_ids = set(stores["Store ID"])

missing_stores = transactions[
    ~transactions["Store ID"].isin(store_ids)
]

print("Transactions with unknown Store ID:",
      len(missing_stores))
```

```python
employee_ids = set(employees["Employee ID"])

missing_employees = transactions[
    ~transactions["Employee ID"].isin(employee_ids)
]

print("Transactions with unknown Employee ID:",
      len(missing_employees))
```
